Создать базовый класс Item в C#, который будет представлять информацию о
товарах, которые могут быть заказаны или возвращены. На основе этого класса

полиморфизма. В каждом из классов должны быть реализованы новые атрибуты и

демонстрации полиморфизма.
Требования к базовому классу Item:
• Атрибуты: ID товара (ItemId), Название (Name), Цена (Price).
• Методы:
o
o GetDetails(): метод для получения детальной информации о товаре.
o CalculateDiscount(): метод для расчета скидки на товар.
o ApplyDiscount(decimal discount): метод для применения скидки к цене
товара.
Требования к производным классам:
1. ЕдиничныйТовар (SingleItem): Должен содержать дополнительные атрибуты,
такие как Единица измерения (UnitMeasure). Метод GetDetails() должен быть
переопределен для добавления информации о единице измерения товара.
2. ПакетныйТовар (PackageItem): Должен содержать дополнительные
атрибуты, такие как Количество единиц в пакете (QuantityPerPackage).
Метод CalculateDiscount() должен быть переопределен для учета количества
единиц в пакете при расчете скидки.
3. СпециальныйТовар (SpecialItem) (если требуется третий класс): Должен
содержать дополнительные атрибуты, такие как Дата истечения скидки
(DiscountExpirationDate). Метод ApplyDiscount() должен быть переопределен
для добавления информации о сроке действия скидки.

```
using System;
using System.Collections.Generic;
using System.Linq;

// Базовый класс Item
public class Item
{
    // Атрибуты с геттерами и сеттерами
    public int ItemId { get; set; }
    public string Name { get; set; }
    public decimal Price { get; set; }
    public string Description { get; set; }
    public DateTime CreatedDate { get; set; }
    public bool IsAvailable { get; set; }
    
    public string Category { get; set; }
    public string Manufacturer { get; set; }
    public double Rating { get; set; }
    public int ViewCount { get; set; }

    // Новые атрибуты
    public string Material { get; set; }
    public string Size { get; set; }
    public bool IsEcoFriendly { get; set; }
    public int WarrantyDays { get; set; }

    // Новые атрибуты
    public int StockQuantity { get; set; }
    public int SoldCount { get; set; }
    public string Color { get; set; }
    public List<string> Tags { get; set; }
    public DateTime LastModified { get; set; }

    // Делегаты и события
    public delegate void ItemEventHandler(object sender, ItemEventArgs e);
    public delegate void PriceChangedEventHandler(decimal oldPrice, decimal newPrice);
    public delegate void StockStatusChangedEventHandler(bool isInStock);

    public event ItemEventHandler ItemUpdated;
    public event PriceChangedEventHandler PriceChanged;
    public event StockStatusChangedEventHandler StockStatusChanged;

    // Конструктор с использованием геттеров и сеттеров
    public Item(int itemId, string name, decimal price, string description)
    {
        ItemId = itemId;
        Name = name;
        Price = price;
        Description = description;
        CreatedDate = DateTime.Now;
        IsAvailable = true;
        Category = "General";
        Manufacturer = "Unknown";
        Rating = 0.0;
        ViewCount = 0;
        Material = "Unknown";
        Size = "Standard";
        IsEcoFriendly = false;
        WarrantyDays = 0;
        
        // Инициализация новых атрибутов
        StockQuantity = 0;
        SoldCount = 0;
        Color = "Unknown";
        Tags = new List<string>();
        LastModified = DateTime.Now;
    }

    // Перегруженный конструктор (ПЕРЕГРУЗКА)
    public Item(int itemId, string name, decimal price, string description, string category, string manufacturer)
        : this(itemId, name, price, description)
    {
        Category = category;
        Manufacturer = manufacturer;
    }

    // Метод для получения детальной информации о товаре
    public virtual string GetDetails()
    {
        return $"ID: {ItemId}, Название: {Name}, Цена: {Price:N2}, Описание: {Description}";
    }

    // Перегруженный метод GetDetails (ПЕРЕГРУЗКА)
    public virtual string GetDetails(bool includeExtendedInfo)
    {
        if (includeExtendedInfo)
        {
            return $"{GetDetails()}, Категория: {Category}, Производитель: {Manufacturer}, Рейтинг: {Rating:F1}";
        }
        return GetDetails();
    }

    // Метод для расчета скидки на товар (базовая реализация - 10%)
    public virtual decimal CalculateDiscount()
    {
        return Price * 0.1m;
    }

    // Перегруженный метод CalculateDiscount (ПЕРЕГРУЗКА)
    public virtual decimal CalculateDiscount(decimal discountPercentage)
    {
        if (discountPercentage < 0 || discountPercentage > 100)
            throw new ArgumentException("Процент скидки должен быть от 0 до 100");
        
        return Price * (discountPercentage / 100m);
    }

    // Метод для применения скидки к цене товара
    public virtual decimal ApplyDiscount(decimal discount)
    {
        if (discount < 0 || discount > Price)
        {
            throw new ArgumentException("Некорректная сумма скидки");
        }
        Price -= discount;
        return Price;
    }

    public virtual void UpdatePrice(decimal newPrice)
    {
        decimal oldPrice = Price;
        Price = newPrice;
        LastModified = DateTime.Now;
        OnPriceChanged(oldPrice, newPrice);
        OnItemUpdated($"Цена товара {Name} обновлена: {newPrice:N2}");
    }

    public virtual string GetCreationInfo()
    {
        return $"Товар создан: {CreatedDate:dd.MM.yyyy HH:mm}";
    }

    public void ToggleAvailability()
    {
        IsAvailable = !IsAvailable;
        Console.WriteLine($"Товар {Name} теперь {(IsAvailable ? "доступен" : "недоступен")}");
    }

    public virtual void UpdateRating(double newRating)
    {
        if (newRating < 0 || newRating > 5)
            throw new ArgumentException("Рейтинг должен быть от 0 до 5");
        
        Rating = newRating;
        Console.WriteLine($"Рейтинг товара {Name} обновлен: {newRating:F1}");
    }

    public virtual void IncrementViewCount()
    {
        ViewCount++;
    }

    public virtual string GetPopularityStatus()
    {
        if (ViewCount > 1000) return "Очень популярный";
        if (ViewCount > 500) return "Популярный";
        if (ViewCount > 100) return "Средняя популярность";
        return "Мало просмотров";
    }

    // Перегруженный метод (ПЕРЕГРУЗКА)
    public virtual string GetPopularityStatus(string prefix)
    {
        return $"{prefix}: {GetPopularityStatus()}";
    }

    // Новые методы
    public virtual string GetMaterialInfo()
    {
        return $"Материал: {Material}, Экологичный: {(IsEcoFriendly ? "Да" : "Нет")}";
    }

    public virtual void UpdateWarranty(int days)
    {
        WarrantyDays = days;
        Console.WriteLine($"Гарантия на {Name} установлена: {days} дней");
    }

    public virtual bool IsUnderWarranty()
    {
        return WarrantyDays > 0 && CreatedDate.AddDays(WarrantyDays) > DateTime.Now;
    }

    public virtual decimal CalculateEcoDiscount()
    {
        return IsEcoFriendly ? Price * 0.05m : 0;
    }

    // Новые методы
    public virtual void UpdateStock(int quantity)
    {
        int oldStock = StockQuantity;
        StockQuantity = quantity;
        LastModified = DateTime.Now;
        
        OnItemUpdated($"Количество на складе изменено с {oldStock} на {quantity}");
        
        if ((oldStock == 0 && quantity > 0) || (oldStock > 0 && quantity == 0))
        {
            StockStatusChanged?.Invoke(quantity > 0);
        }
    }

    public virtual void SellItem(int quantity = 1)
    {
        if (quantity > StockQuantity)
            throw new InvalidOperationException("Недостаточно товара на складе");

        StockQuantity -= quantity;
        SoldCount += quantity;
        LastModified = DateTime.Now;
        
        OnItemUpdated($"Продано {quantity} единиц товара");
    }

    public virtual void AddTag(string tag)
    {
        Tags ??= new List<string>();
        if (!Tags.Contains(tag))
        {
            Tags.Add(tag);
            OnItemUpdated($"Добавлен тег: {tag}");
        }
    }

    public virtual bool HasTag(string tag)
    {
        return Tags?.Contains(tag) ?? false;
    }

    public virtual decimal CalculateProfit()
    {
        return SoldCount * Price;
    }

    // Методы для работы с событиями
    protected virtual void OnItemUpdated(string message)
    {
        ItemUpdated?.Invoke(this, new ItemEventArgs(message));
    }

    protected virtual void OnPriceChanged(decimal oldPrice, decimal newPrice)
    {
        PriceChanged?.Invoke(oldPrice, newPrice);
    }
}

// Аргументы события
public class ItemEventArgs : EventArgs
{
    public string Message { get; }
    public DateTime EventTime { get; }

    public ItemEventArgs(string message)
    {
        Message = message;
        EventTime = DateTime.Now;
    }
}

// Интерфейсы для явной реализации
public interface IReviewable
{
    void AddReview(string review, int rating);
    List<string> GetReviews();
    double CalculateAverageRating();
}

public interface ITaxable
{
    decimal CalculateTax();
    string GetTaxInfo();
}

// Generic класс для работы с коллекциями товаров
public class ItemCollection<T> where T : Item
{
    private List<T> items;

    // Событие для коллекции
    public event Action<string, T> CollectionModified;

    public ItemCollection()
    {
        items = new List<T>();
    }

    public void AddItem(T item)
    {
        items.Add(item);
        CollectionModified?.Invoke("Добавлен", item);
        
        // Подписываемся на события товара
        item.ItemUpdated += (sender, e) => 
            Console.WriteLine($"[{item.Name}] {e.Message} в {e.EventTime:HH:mm:ss}");
    }

    public bool RemoveItem(int itemId)
    {
        var item = items.FirstOrDefault(i => i.ItemId == itemId);
        if (item != null)
        {
            items.Remove(item);
            CollectionModified?.Invoke("Удален", item);
            return true;
        }
        return false;
    }

    public T FindItemById(int itemId)
    {
        return items.FirstOrDefault(i => i.ItemId == itemId);
    }

    public List<T> FindItemsByCategory(string category)
    {
        return items.Where(i => i.Category.Equals(category, StringComparison.OrdinalIgnoreCase)).ToList();
    }

    public List<T> GetAvailableItems()
    {
        return items.Where(i => i.IsAvailable).ToList();
    }

    public List<T> GetItemsSortedByPrice(bool ascending = true)
    {
        return ascending ? items.OrderBy(i => i.Price).ToList() : items.OrderByDescending(i => i.Price).ToList();
    }

    public decimal CalculateTotalValue()
    {
        return items.Sum(i => i.Price);
    }

    public void DisplayAllItems()
    {
        Console.WriteLine($"\nВсе товары в коллекции ({typeof(T).Name}):");
        foreach (var item in items)
        {
            Console.WriteLine(item.GetDetails(true));
        }
    }

    // Новые методы для работы с коллекцией
    public List<T> GetLowStockItems(int threshold = 5)
    {
        return items.Where(i => i.StockQuantity <= threshold).ToList();
    }

    public List<T> GetBestSellers(int topN = 5)
    {
        return items.OrderByDescending(i => i.SoldCount).Take(topN).ToList();
    }

    public List<T> FindItemsByTag(string tag)
    {
        return items.Where(i => i.HasTag(tag)).ToList();
    }

    public Dictionary<string, decimal> GetProfitByCategory()
    {
        return items.GroupBy(i => i.Category)
                   .ToDictionary(g => g.Key, g => g.Sum(i => i.CalculateProfit()));
    }

    public void ApplyToAll(Action<T> action)
    {
        foreach (var item in items)
        {
            action(item);
        }
    }
}

// Класс для управления зависимостями
public class ItemService
{
    private readonly Item _item;

    public ItemService(Item item)
    {
        _item = item;
    }

    public void ProcessItem()
    {
        Console.WriteLine($"Обработка товара: {_item.Name}");
        _item.IncrementViewCount();
        Console.WriteLine(_item.GetDetails(true));
        Console.WriteLine($"Статус популярности: {_item.GetPopularityStatus()}");
        
        if (_item is ITaxable taxable)
        {
            Console.WriteLine($"Налог: {taxable.CalculateTax():N2}");
        }
    }
}

// Производный класс ЕдиничныйТовар с явной реализацией интерфейсов
public class SingleItem : Item, IReviewable, ITaxable
{
    // Дополнительный атрибут с геттером и сеттером
    public string UnitMeasure { get; set; }
    public string Brand { get; set; }
    public string Barcode { get; set; }
    
    public DateTime ExpiryDate { get; set; }
    public string StorageConditions { get; set; }
    public int MinOrderQuantity { get; set; }

    // Новые атрибуты
    private List<string> _reviews;
    private List<int> _ratings;

    // Новые атрибуты
    public string SerialNumber { get; set; }
    public string BatchNumber { get; set; }
    public bool IsPerishable { get; set; }
    public DateTime ProductionDate { get; set; }

    // Конструктор с использованием геттеров и сеттеров
    public SingleItem(int itemId, string name, decimal price, string description, string unitMeasure, string brand) 
        : base(itemId, name, price, description)
    {
        UnitMeasure = unitMeasure;
        Brand = brand;
        Barcode = GenerateBarcode();
        ExpiryDate = DateTime.Now.AddDays(30);
        StorageConditions = "Комнатная температура";
        MinOrderQuantity = 1;
        _reviews = new List<string>();
        _ratings = new List<int>();

        // Инициализация новых атрибутов
        SerialNumber = string.Empty;
        BatchNumber = string.Empty;
        IsPerishable = false;
        ProductionDate = DateTime.Now;
    }

    // Явная реализация интерфейса IReviewable
    void IReviewable.AddReview(string review, int rating)
    {
        _reviews.Add(review);
        _ratings.Add(rating);
        UpdateRating(_ratings.Average());
    }

    List<string> IReviewable.GetReviews()
    {
        return _reviews;
    }

    double IReviewable.CalculateAverageRating()
    {
        return _ratings.Count > 0 ? _ratings.Average() : 0;
    }

    // Явная реализация интерфейса ITaxable
    decimal ITaxable.CalculateTax()
    {
        return Price * 0.2m; // 20% налог
    }

    string ITaxable.GetTaxInfo()
    {
        return $"Налог на товар: {((ITaxable)this).CalculateTax():N2}";
    }

    // Переопределение метода GetDetails() (ПЕРЕОПРЕДЕЛЕНИЕ)
    public override string GetDetails()
    {
        return base.GetDetails() + $", Единица измерения: {UnitMeasure}, Бренд: {Brand}";
    }

    // Переопределение с расширенной информацией (ПЕРЕОПРЕДЕЛЕНИЕ)
    public override string GetDetails(bool includeExtendedInfo)
    {
        if (includeExtendedInfo)
        {
            return $"{GetDetails()}, Срок годности: {ExpiryDate:dd.MM.yyyy}, Условия хранения: {StorageConditions}";
        }
        return GetDetails();
    }

    public string GenerateBarcode()
    {
        return $"{ItemId:0000}{DateTime.Now:MMddHHmm}";
    }

    public void UpdateBrand(string newBrand)
    {
        Brand = newBrand;
        Console.WriteLine($"Бренд товара {Name} обновлен: {newBrand}");
    }

    // Новые методы
    public override decimal CalculateEcoDiscount()
    {
        decimal baseDiscount = base.CalculateEcoDiscount();
        return IsEcoFriendly ? baseDiscount + (Price * 0.03m) : baseDiscount;
    }

    public string GetExpiryStatus()
    {
        int daysLeft = (ExpiryDate - DateTime.Now).Days;
        return daysLeft > 7 ? "Свежий" : daysLeft > 0 ? "Скоро истекает" : "Просрочен";
    }

    // Новые методы
    public override void UpdateStock(int quantity)
    {
        base.UpdateStock(quantity);
        if (IsPerishable && quantity > 0)
        {
            CheckExpiryStatus();
        }
    }

    public void CheckExpiryStatus()
    {
        if (IsPerishable && DateTime.Now > ExpiryDate)
        {
            StockQuantity = 0;
            OnItemUpdated("Товар просрочен и снят с продажи");
        }
    }

    public string GetBatchInfo()
    {
        return $"Партия: {BatchNumber}, Серия: {SerialNumber}, Произведено: {ProductionDate:dd.MM.yyyy}";
    }

    public override decimal CalculateProfit()
    {
        decimal baseProfit = base.CalculateProfit();
        return IsEcoFriendly ? baseProfit * 1.1m : baseProfit;
    }
}

// Производный класс ПакетныйТовар
public class PackageItem : Item, ITaxable
{
    public int ItemsInPackage { get; set; }
    public string PackageType { get; set; }
    public decimal PackageWeight { get; set; }

    // Новые атрибуты
    public bool IsBulk { get; set; }
    public int MaxStackCount { get; set; }

    // Новые атрибуты
    public bool IsFragile { get; set; }
    public string PackagingColor { get; set; }
    public int ReorderLevel { get; set; }
    public List<string> PackageFeatures { get; set; }

    public PackageItem(int itemId, string name, decimal price, string description, int itemsInPackage, string packageType)
        : base(itemId, name, price, description)
    {
        ItemsInPackage = itemsInPackage;
        PackageType = packageType;
        PackageWeight = itemsInPackage * 0.1m;
        IsBulk = itemsInPackage > 10;
        MaxStackCount = 5;

        // Инициализация новых атрибутов
        IsFragile = false;
        PackagingColor = "Standard";
        ReorderLevel = 10;
        PackageFeatures = new List<string>();
    }

    // Явная реализация интерфейса ITaxable
    decimal ITaxable.CalculateTax()
    {
        return Price * (IsBulk ? 0.15m : 0.18m); // Разный налог для опта и розницы
    }

    string ITaxable.GetTaxInfo()
    {
        return $"Налог на упаковку: {((ITaxable)this).CalculateTax():N2}, Оптовый: {(IsBulk ? "Да" : "Нет")}";
    }

    // Новые методы
    public override string GetDetails()
    {
        return base.GetDetails() + $", Количество в упаковке: {ItemsInPackage}, Тип упаковки: {PackageType}";
    }

    public decimal GetPricePerItem()
    {
        return Price / ItemsInPackage;
    }

    public override void UpdatePrice(decimal newPrice)
    {
        base.UpdatePrice(newPrice);
        Console.WriteLine($"Цена за единицу: {GetPricePerItem():N2}");
    }

    public bool CanStackWith(PackageItem other)
    {
        return this.PackageType == other.PackageType && this.MaxStackCount == other.MaxStackCount;
    }

    // Новые методы
    public override void UpdateStock(int quantity)
    {
        base.UpdateStock(quantity);
        
        if (quantity <= ReorderLevel && quantity > 0)
        {
            OnItemUpdated($"Внимание! Запас товара на низком уровне: {quantity}");
        }
    }

    public void AddPackageFeature(string feature)
    {
        PackageFeatures ??= new List<string>();
        if (!PackageFeatures.Contains(feature))
        {
            PackageFeatures.Add(feature);
            OnItemUpdated($"Добавлена особенность упаковки: {feature}");
        }
    }

    public bool NeedsRestocking()
    {
        return StockQuantity <= ReorderLevel;
    }

    public override decimal CalculateProfit()
    {
        return base.CalculateProfit() * (IsBulk ? 0.9m : 1.0m);
    }
}

// Производный класс ЦифровойТовар
public class DigitalItem : Item, IReviewable
{
    public string FileFormat { get; set; }
    public decimal FileSize { get; set; }
    public string DownloadLink { get; set; }

    // Новые атрибуты
    private List<string> _reviews;
    private List<int> _ratings;
    public bool HasDRM { get; set; }

    // Новые атрибуты
    public string LicenseType { get; set; }
    public int DownloadCount { get; set; }
    public string Version { get; set; }
    public List<string> SupportedPlatforms { get; set; }

    public DigitalItem(int itemId, string name, decimal price, string description, string fileFormat, decimal fileSize)
        : base(itemId, name, price, description)
    {
        FileFormat = fileFormat;
        FileSize = fileSize;
        DownloadLink = $"https://store.com/download/{itemId}";
        _reviews = new List<string>();
        _ratings = new List<int>();
        HasDRM = true;
        IsEcoFriendly = true; // Цифровые товары всегда экологичны

        // Инициализация новых атрибутов
        LicenseType = "Personal";
        DownloadCount = 0;
        Version = "1.0";
        SupportedPlatforms = new List<string>();
    }

    // Явная реализация интерфейса IReviewable
    void IReviewable.AddReview(string review, int rating)
    {
        _reviews.Add(review);
        _ratings.Add(rating);
        UpdateRating(_ratings.Average());
    }

    List<string> IReviewable.GetReviews()
    {
        return _reviews;
    }

    double IReviewable.CalculateAverageRating()
    {
        return _ratings.Count > 0 ? _ratings.Average() : 0;
    }

    // Новые методы
    public override string GetDetails()
    {
        return base.GetDetails() + $", Формат: {FileFormat}, Размер: {FileSize}MB";
    }

    public override decimal CalculateDiscount()
    {
        return base.CalculateDiscount() + (Price * 0.05m); // Дополнительная скидка для цифровых товаров
    }

    public string GetDownloadInfo()
    {
        return $"Ссылка для скачивания: {DownloadLink}, DRM: {(HasDRM ? "Да" : "Нет")}";
    }

    public override decimal CalculateEcoDiscount()
    {
        return Price * 0.1m; // Большая скидка за экологичность
    }

    // Новые методы
    public void IncrementDownloadCount()
    {
        DownloadCount++;
        SoldCount++;
        OnItemUpdated($"Файл скачан. Всего скачиваний: {DownloadCount}");
    }

    public void AddSupportedPlatform(string platform)
    {
        SupportedPlatforms ??= new List<string>();
        if (!SupportedPlatforms.Contains(platform))
        {
            SupportedPlatforms.Add(platform);
            OnItemUpdated($"Добавлена поддержка платформы: {platform}");
        }
    }

    public string GetVersionInfo()
    {
        return $"Версия: {Version}, Лицензия: {LicenseType}, Скачиваний: {DownloadCount}";
    }

    public override decimal CalculateProfit()
    {
        return base.CalculateProfit() * 1.2m; // Цифровые товары имеют более высокую маржу
    }
}

// Создание товаров
SingleItem milk = new SingleItem(1, "Молоко Простоквашино", 80.50m, "Парное молоко 3.2%", "литр", "Простоквашино");
PackageItem pencils = new PackageItem(2, "Цветные карандаши", 150.00m, "Набор цветных карандашей", 12, "картонная коробка");
DigitalItem ebook = new DigitalItem(3, "C# Programming Guide", 500.00m, "Электронная книга по C#", "PDF", 5.2m);

// Настройка свойств
milk.Material = "Пластик";
milk.IsEcoFriendly = true;
milk.WarrantyDays = 14;

pencils.IsEcoFriendly = true;
pencils.WarrantyDays = 30;

// Настройка новых свойств
milk.StockQuantity = 50;
milk.Color = "Белый";
milk.Tags = new List<string> { "молочка", "охлажденный" };
milk.BatchNumber = "BATCH001";
milk.SerialNumber = "SN123456";
milk.IsPerishable = true;
milk.ProductionDate = DateTime.Now.AddDays(-2);

pencils.StockQuantity = 100;
pencils.Color = "Разноцветный";
pencils.Tags = new List<string> { "канцелярия", "творчество" };
pencils.IsFragile = true;
pencils.PackagingColor = "Желтый";
pencils.ReorderLevel = 20;

ebook.StockQuantity = 9999;
ebook.Color = "Синий";
ebook.Tags = new List<string> { "образование", "программирование" };
ebook.LicenseType = "Personal";
ebook.Version = "1.0";
ebook.SupportedPlatforms = new List<string> { "Windows", "macOS" };

// Использование явной реализации интерфейсов
((IReviewable)milk).AddReview("Отличное молоко!", 5);
((IReviewable)milk).AddReview("Свежий продукт", 4);

((IReviewable)ebook).AddReview("Полезная книга", 5);
((IReviewable)ebook).AddReview("Хорошие примеры", 4);

// Создание и использование коллекций
var singleItems = new ItemCollection<SingleItem>();
var packageItems = new ItemCollection<PackageItem>();
var digitalItems = new ItemCollection<DigitalItem>();

// Подписка на события коллекций
singleItems.CollectionModified += (action, item) => 
    Console.WriteLine($"[SingleItems] {action}: {item.Name}");

packageItems.CollectionModified += (action, item) => 
    Console.WriteLine($"[PackageItems] {action}: {item.Name}");

digitalItems.CollectionModified += (action, item) => 
    Console.WriteLine($"[DigitalItems] {action}: {item.Name}");

// Подписка на события товаров
milk.PriceChanged += (oldPrice, newPrice) => 
    Console.WriteLine($"Цена молока изменилась: {oldPrice:N2} → {newPrice:N2}");

milk.StockStatusChanged += (isInStock) => 
    Console.WriteLine($"Статус наличия молока: {(isInStock ? "В наличии" : "Нет в наличии")}");

// Добавление в коллекции
singleItems.AddItem(milk);
packageItems.AddItem(pencils);
digitalItems.AddItem(ebook);

// Управление зависимостями
ItemService milkService = new ItemService(milk);
ItemService pencilsService = new ItemService(pencils);
ItemService ebookService = new ItemService(ebook);

milkService.ProcessItem();
Console.WriteLine();
pencilsService.ProcessItem();
Console.WriteLine();
ebookService.ProcessItem();

Console.WriteLine(((ITaxable)milk).GetTaxInfo());
Console.WriteLine(((ITaxable)pencils).GetTaxInfo());

Console.WriteLine($"Средний рейтинг молока: {((IReviewable)milk).CalculateAverageRating():F1}");
Console.WriteLine($"Средний рейтинг книги: {((IReviewable)ebook).CalculateAverageRating():F1}");

Console.WriteLine($"Эко-скидка на молоко: {milk.CalculateEcoDiscount():N2}");
Console.WriteLine($"Эко-скидка на карандаши: {pencils.CalculateEcoDiscount():N2}");
Console.WriteLine($"Эко-скидка на книгу: {ebook.CalculateEcoDiscount():N2}");

Console.WriteLine(milk.GetMaterialInfo());
Console.WriteLine($"Статус годности молока: {milk.GetExpiryStatus()}");
Console.WriteLine(ebook.GetDownloadInfo());
Console.WriteLine($"Цена за карандаш: {pencils.GetPricePerItem():N2}");

// Проверка гарантии
Console.WriteLine($"Молоко на гарантии: {milk.IsUnderWarranty()}");
Console.WriteLine($"Карандаши на гарантии: {pencils.IsUnderWarranty()}");

// Демонстрация работы новых методов
milk.SellItem(5);
pencils.SellItem(3);
ebook.IncrementDownloadCount();

milk.AddTag("популярный");
pencils.AddPackageFeature("Перерабатываемая упаковка");
ebook.AddSupportedPlatform("Linux");

milk.UpdatePrice(85.00m);
pencils.UpdateStock(15);

// Работа с коллекциями
var lowStockItems = packageItems.GetLowStockItems(20);
var bestSellers = singleItems.GetBestSellers(3);
var educationItems = digitalItems.FindItemsByTag("образование");
var profitByCategory = singleItems.GetProfitByCategory();

// Применение делегата ко всем элементам коллекции
singleItems.ApplyToAll(item => item.AddTag("акционный"));

Console.WriteLine($"Прибыль по категориям: {string.Join(", ", profitByCategory)}");
Console.WriteLine($"Низкий запас: {lowStockItems.Count} товаров");
Console.WriteLine($"Информация о партии молока: {milk.GetBatchInfo()}");
Console.WriteLine($"Информация о версии книги: {ebook.GetVersionInfo()}");
Console.WriteLine($"Нужно пополнить карандаши: {pencils.NeedsRestocking()}");
```
