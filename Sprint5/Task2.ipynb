<h1 style="color:DodgerBlue">Делегаты, лямбда-выражения и события<T></h1>



Делегаты, лямбда-выражения и события являются важными концепциями в C#, которые позволяют работать с методами как с объектами, писать краткий код и управлять реакциями на события. Рассмотрим их на примере гипотетических классов `Transport`, `Car`, и `Bike`.

### Делегаты

**Делегат** — это тип безопасного указателя на метод, похожий на функцию в C или C++. Делегаты в C# используются для передачи методов как параметров, а также для создания обратных вызовов.

**Пример:**

Создадим делегат, который может ссылаться на методы, выводящие информацию о транспортном средстве:

```csharp
public delegate void DisplayDelegate();

public class Transport
{
    public string Model { get; set; }
    public string Manufacturer { get; set; }

    public Transport(string model, string manufacturer)
    {
        Model = model;
        Manufacturer = manufacturer;
    }

    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Model: {Model}, Manufacturer: {Manufacturer}");
    }
}
```

### Лямбда-выражения

**Лямбда-выражения** позволяют вам писать анонимные методы в более краткой форме. Это часто используется в LINQ и асинхронном программировании.

**Пример:**

```csharp
using System;
using System.Collections.Generic;
using System.Linq;


 
List<Transport> transports = new List<Transport>
{
    new Car("Model S", "Tesla", 4),
    new Bike("Speedster", "Giant", true),
    new Car("Model 3", "Tesla", 4),
};

// Используем лямбда-выражение для фильтрации транспортных средств
var teslaCars = transports.OfType<Car>().Where(c => c.Manufacturer == "Tesla");

foreach (var car in teslaCars)
{
    car.DisplayInfo();
}


```

### События

**События** используются для оповещения подписчиков о случившихся действиях, предоставляя более высокоуровневый способ взаимодействия между различными частями кода.

**Пример:**

```csharp
public class Transport
{
    public string Model { get; set; }
    public string Manufacturer { get; set; }

    public delegate void TransportHandler(string message);
    public event TransportHandler TransportUpdated;

    public void UpdateModel(string newModel)
    {
        Model = newModel;
        // Вызываем событие после обновления модели
        TransportUpdated?.Invoke($"Model updated to {newModel}");
    }
}



Transport transport = new Transport { Model = "Old Model", Manufacturer = "Generic" };
// Подписываем метод на событие
transport.TransportUpdated += OnTransportUpdated;
// Обновляем модель; это вызовет событие
transport.UpdateModel("New Model");


static void OnTransportUpdated(string message)
{
    Console.WriteLine(message);
}
```

### Объяснения и комментарии:

- **Делегаты**: Тип данных, который можно использовать для хранения ссылок на методы. Полезно для обратных вызовов и передачи методов.

- **Лямбда-выражения**: Более краткая запись анонимных методов, часто используется в LINQ для фильтрации, проекций и агрегаций.

- **События**: Механизм уведомления объектов об изменениях состояния или других действиях. Позволяют подписываться на события в одной части программы и реакцию на них в другой.

Эти механизмы делают программу гибкой и более структурированной, обеспечивая четкое отделение логики и упрощая повторное использование кода.

````
// Делегат для отображения информации
public delegate void DisplayDelegate();

// Делегат для событий транспорта
public delegate void TransportHandler(string message);

public class Transport
{
    public string Model { get; set; }
    public string Manufacturer { get; set; }

    // Событие для уведомлений
    public event TransportHandler TransportUpdated;

    public Transport(string model, string manufacturer)
    {
        Model = model;
        Manufacturer = manufacturer;
    }

    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Модель: {Model}, Производитель: {Manufacturer}");
    }

    // Метод для вызова события
    protected virtual void OnTransportUpdated(string message)
    {
        TransportUpdated?.Invoke(message);
    }

    public void UpdateModel(string newModel)
    {
        string oldModel = Model;
        Model = newModel;
        OnTransportUpdated($"Модель изменена с '{oldModel}' на '{newModel}'");
    }
}

public class Car : Transport
{
    public int NumberOfDoors { get; set; }

    public Car(string model, string manufacturer, int numberOfDoors)
        : base(model, manufacturer)
    {
        NumberOfDoors = numberOfDoors;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Количество дверей: {NumberOfDoors}");
    }

    public void Accelerate()
    {
        OnTransportUpdated($"Автомобиль {Model} ускоряется!");
    }
}

public class Bike : Transport
{
    public bool HasCarrier { get; set; }

    public Bike(string model, string manufacturer, bool hasCarrier)
        : base(model, manufacturer)
    {
        HasCarrier = hasCarrier;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Имеет багажник: {HasCarrier}");
    }

    public void RingBell()
    {
        OnTransportUpdated($"Велосипед {Model} звонит в звонок!");
    }
}

// Использование делегатов
Car car = new Car("Model S", "Tesla", 4);
Bike bike = new Bike("Speedster", "Giant", true);

DisplayDelegate displayCar = car.DisplayInfo;
DisplayDelegate displayBike = bike.DisplayInfo;

Console.WriteLine("Использование делегатов:");
displayCar();
displayBike();
Console.WriteLine();

// Использование лямбда-выражений
List<Transport> transports = new List<Transport>
{
    new Car("Model S", "Tesla", 4),
    new Bike("Speedster", "Giant", true),
    new Car("Model 3", "Tesla", 4),
    new Bike("Mountain", "Trek", false),
    new Car("Camry", "Toyota", 4)
};

Console.WriteLine("Использование лямбда-выражений:");

// Фильтрация Tesla автомобилей
var teslaCars = transports.OfType<Car>().Where(c => c.Manufacturer == "Tesla");
Console.WriteLine("Автомобили Tesla:");
foreach (var teslaCar in teslaCars)
{
    teslaCar.DisplayInfo();
}

Console.WriteLine();

// Фильтрация велосипедов с багажником
var bikesWithCarrier = transports.OfType<Bike>().Where(b => b.HasCarrier);
Console.WriteLine("Велосипеды с багажником:");
foreach (var bikeWithCarrier in bikesWithCarrier)
{
    bikeWithCarrier.DisplayInfo();
}

Console.WriteLine();

// Сортировка по модели
var sortedTransports = transports.OrderBy(t => t.Model);
Console.WriteLine("Транспорт отсортированный по модели:");
foreach (var transport in sortedTransports)
{
    transport.DisplayInfo();
}

Console.WriteLine();

// Использование событий
Console.WriteLine("Использование событий:");

// Подписка на события
car.TransportUpdated += (message) => Console.WriteLine($"Событие автомобиля: {message}");
bike.TransportUpdated += (message) => Console.WriteLine($"Событие велосипеда: {message}");

// Вызов методов, которые генерируют события
car.UpdateModel("Model X");
car.Accelerate();

bike.UpdateModel("City Explorer");
bike.RingBell();

// Групповое использование делегатов
Console.WriteLine("\nГрупповые делегаты:");
DisplayDelegate multiDisplay = displayCar + displayBike;
multiDisplay();

// Лямбда с действиями
Console.WriteLine("\nЛямбда с действиями:");
Action<string> printAction = (text) => Console.WriteLine($"Действие: {text}");
printAction("Привет от действия!");

// Функции с лямбдами
Func<int, int, string> sumFunc = (a, b) => $"Сумма: {a + b}";
Console.WriteLine(sumFunc(5, 3));

// Дополнительные примеры с лямбдами
Console.WriteLine("\nДополнительные примеры с лямбдами:");

// Поиск автомобилей с 4 дверями
var fourDoorCars = transports.OfType<Car>().Where(c => c.NumberOfDoors == 4);
Console.WriteLine("Автомобили с 4 дверями:");
foreach (var fourDoorCar in fourDoorCars)
{
    fourDoorCar.DisplayInfo();
}

// Подсчет общего количества транспорта
int totalTransports = transports.Count;
Console.WriteLine($"Всего транспортных средств: {totalTransports}");

// Проверка наличия определенного производителя
bool hasToyota = transports.Any(t => t.Manufacturer == "Toyota");
Console.WriteLine($"Есть ли Toyota в списке: {hasToyota}");
````

````
// Делегат для отображения информации о животных
public delegate void AnimalDisplayDelegate();

// Делегат для событий животных
public delegate void AnimalEventHandler(string message);

public class Animal
{
    public string Name { get; set; }
    public int Age { get; set; }

    // Событие для уведомлений
    public event AnimalEventHandler AnimalEvent;

    public Animal(string name, int age)
    {
        Name = name;
        Age = age;
    }

    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Имя: {Name}, Возраст: {Age} лет");
    }

    public virtual void MakeSound()
    {
        Console.WriteLine($"{Name} издает звук");
    }

    // Метод для вызова события
    protected virtual void OnAnimalEvent(string message)
    {
        AnimalEvent?.Invoke(message);
    }

    public void UpdateName(string newName)
    {
        string oldName = Name;
        Name = newName;
        OnAnimalEvent($"Имя изменено с '{oldName}' на '{newName}'");
    }
}

public class Dog : Animal
{
    public string Breed { get; set; }

    public Dog(string name, int age, string breed)
        : base(name, age)
    {
        Breed = breed;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Порода: {Breed}");
    }

    public override void MakeSound()
    {
        Console.WriteLine($"{Name} говорит: Гав-гав!");
    }

    public void FetchBall()
    {
        OnAnimalEvent($"Собака {Name} принесла мяч!");
    }
}

public class Cat : Animal
{
    public bool IsIndoor { get; set; }

    public Cat(string name, int age, bool isIndoor)
        : base(name, age)
    {
        IsIndoor = isIndoor;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Домашний кот: {IsIndoor}");
    }

    public override void MakeSound()
    {
        Console.WriteLine($"{Name} говорит: Мяу-мяу!");
    }

    public void ClimbTree()
    {
        OnAnimalEvent($"Кот {Name} залез на дерево!");
    }
}

public class Bird : Animal
{
    public double Wingspan { get; set; }

    public Bird(string name, int age, double wingspan)
        : base(name, age)
    {
        Wingspan = wingspan;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Размах крыльев: {Wingspan} см");
    }

    public override void MakeSound()
    {
        Console.WriteLine($"{Name} говорит: Чик-чирик!");
    }

    public void Fly()
    {
        OnAnimalEvent($"Птица {Name} взлетела в небо!");
    }
}

public class Fish : Animal
{
    public string WaterType { get; set; }

    public Fish(string name, int age, string waterType)
        : base(name, age)
    {
        WaterType = waterType;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Тип воды: {WaterType}");
    }

    public override void MakeSound()
    {
        Console.WriteLine($"{Name} пускает пузыри!");
    }

    public void Swim()
    {
        OnAnimalEvent($"Рыбка {Name} поплыла!");
    }
}

// Использование делегатов
Dog dog = new Dog("Барсик", 3, "Лабрадор");
Cat cat = new Cat("Мурка", 2, true);
Bird bird = new Bird("Кеша", 1, 15.5);
Fish fish = new Fish("Немо", 1, "Морская");

AnimalDisplayDelegate displayDog = dog.DisplayInfo;
AnimalDisplayDelegate displayCat = cat.DisplayInfo;
AnimalDisplayDelegate displayBird = bird.DisplayInfo;
AnimalDisplayDelegate displayFish = fish.DisplayInfo;

Console.WriteLine("ИСПОЛЬЗОВАНИЕ ДЕЛЕГАТОВ:");
displayDog();
displayCat();
displayBird();
displayFish();
Console.WriteLine();

// Использование лямбда-выражений
List<Animal> animals = new List<Animal>
{
    new Dog("Барсик", 3, "Лабрадор"),
    new Dog("Рекс", 4, "Овчарка"),
    new Cat("Мурка", 2, true),
    new Cat("Васька", 1, false),
    new Bird("Кеша", 1, 15.5),
    new Bird("Гоша", 2, 12.0),
    new Fish("Немо", 1, "Морская"),
    new Fish("Золотик", 3, "Пресная")
};

Console.WriteLine("ИСПОЛЬЗОВАНИЕ ЛЯМБДА-ВЫРАЖЕНИЙ:");

// Фильтрация собак определенной породы
var labradors = animals.OfType<Dog>().Where(d => d.Breed == "Лабрадор");
Console.WriteLine("Собаки породы Лабрадор:");
foreach (var labrador in labradors)
{
    labrador.DisplayInfo();
}

Console.WriteLine();

// Фильтрация домашних котов
var indoorCats = animals.OfType<Cat>().Where(c => c.IsIndoor);
Console.WriteLine("Домашние коты:");
foreach (var indoorCat in indoorCats)
{
    indoorCat.DisplayInfo();
}

Console.WriteLine();

// Сортировка животных по возрасту
var sortedByAge = animals.OrderBy(a => a.Age);
Console.WriteLine("Животные отсортированные по возрасту:");
foreach (var animal in sortedByAge)
{
    animal.DisplayInfo();
}

Console.WriteLine();

// Поиск животных старше 2 лет
var olderAnimals = animals.Where(a => a.Age > 2);
Console.WriteLine("Животные старше 2 лет:");
foreach (var olderAnimal in olderAnimals)
{
    olderAnimal.DisplayInfo();
}

Console.WriteLine();

// Использование событий
Console.WriteLine("ИСПОЛЬЗОВАНИЕ СОБЫТИЙ:");

// Подписка на события с помощью лямбда-выражений
dog.AnimalEvent += (message) => Console.WriteLine($"СОБАКА: {message}");
cat.AnimalEvent += (message) => Console.WriteLine($"КОТ: {message}");
bird.AnimalEvent += (message) => Console.WriteLine($"ПТИЦА: {message}");
fish.AnimalEvent += (message) => Console.WriteLine($"РЫБА: {message}");

// Вызов методов, которые генерируют события
dog.UpdateName("Шарик");
dog.FetchBall();

cat.UpdateName("Пушистик");
cat.ClimbTree();

bird.UpdateName("Каркуша");
bird.Fly();

fish.UpdateName("Бульбуль");
fish.Swim();

// Групповое использование делегатов
Console.WriteLine("\nГРУППОВЫЕ ДЕЛЕГАТЫ:");
AnimalDisplayDelegate multiDisplay = displayDog + displayCat + displayBird + displayFish;
multiDisplay();

// Лямбда с действиями
Console.WriteLine("\nЛЯМБДА С ДЕЙСТВИЯМИ:");
Action<string> animalAction = (text) => Console.WriteLine($"Действие животного: {text}");
animalAction("Все животные сыты и довольны!");

// Функции с лямбдами
Console.WriteLine("\nФУНКЦИИ С ЛЯМБДАМИ:");
Func<int, int, string> ageSumFunc = (age1, age2) => $"Сумма возрастов: {age1 + age2} лет";
Console.WriteLine(ageSumFunc(3, 2));

// Дополнительные примеры с лямбдами
Console.WriteLine("\nДОПОЛНИТЕЛЬНЫЕ ПРИМЕРЫ С ЛЯМБДАМИ:");

// Подсчет животных каждого типа
int dogCount = animals.OfType<Dog>().Count();
int catCount = animals.OfType<Cat>().Count();
int birdCount = animals.OfType<Bird>().Count();
int fishCount = animals.OfType<Fish>().Count();

Console.WriteLine($"Всего собак: {dogCount}");
Console.WriteLine($"Всего котов: {catCount}");
Console.WriteLine($"Всего птиц: {birdCount}");
Console.WriteLine($"Всего рыб: {fishCount}");

// Поиск самого старого животного
var oldestAnimal = animals.OrderByDescending(a => a.Age).First();
Console.WriteLine($"\nСамое старое животное:");
oldestAnimal.DisplayInfo();

// Группировка животных по типу
var groupedAnimals = animals.GroupBy(a => a.GetType().Name);
Console.WriteLine("\nЖивотные сгруппированные по типу:");
foreach (var group in groupedAnimals)
{
    Console.WriteLine($"{group.Key}: {group.Count()} животных");
}
````
